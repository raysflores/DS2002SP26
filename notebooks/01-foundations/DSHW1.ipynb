Create tables with:

- Foreign keys

Replace all TODOs below.


```python

# TODO: Drop tables if they exist, then CREATE all required tables

exec_sql('''

DROP TABLE IF EXISTS listens;
DROP TABLE IF EXISTS tracks;
DROP TABLE IF EXISTS albums;
DROP TABLE IF EXISTS artists;
DROP TABLE IF EXISTS users;


CREATE TABLE users ( 
    user_id INTERGER PRIMARY KEY, 
    name TEXT NOT NULL 
);

CREATE TABLE artists ( 
    artist_id INTERGER PRIMARY KEY, 
    name TEXT NOT NULL
);

CREATE TABLE albums ( 
    album_id INTERGER PRIMARY KEY, 
    title TEXT NOT NULL, 
    artist_id INTEGER NOT NULL,
    release_year INTEGER, 
    FOREIGN KEY (artist_id) REFERENCES artists(artist_id)
);

CREATE TABLE tracks ( 
    track_id INTERGER PRIMARY KEY, 
    title TEXT NOT NULL, 
    album_id INTEGER NOT NULL,
    duration_seconds INTEGER, 
    FOREIGN KEY (album_id) REFERENCES albums(album_id)
);

CREATE TABLE listens ( 
    listen_id INTEGER PRIMARY KEY,
    user_id INTEGER, 
    track_id INTEGER,
    listen_time TEXT
);

''')

print("Tables created.")
```

### Q1. List all users.

### Q3. List all albums with their artist name.

```python

run_or_todo('''
SELECT albums.title, artists.name
FROM albums
JOIN artists
ON albums.artist_id = artists.artist_id;
''', label="Q3")
```

### Q4. List all tracks with their album title and artist name.

```python

run_or_todo('''
SELECT tracks.title, albums.title, artists.name
FROM tracks
JOIN albums
ON tracks.album_id = albums.album_id
JOIN artists
ON albums.artist_id = artists.artist_id;
''', label="Q4")
```

### Q5. Show all listening events with user name, track title, and timestamp.

```python

run_or_todo('''
SELECT users.name, tracks.title, listens.listen_time
FROM listens
JOIN users ON listens.user_id = users.user_id
JOIN tracks ON listens.track_id = tracks.track_id;
''', label="Q5")
```

### Q6. How many tracks does each album contain?

```python

run_or_todo('''
SELECT albums.title, COUNT(tracks.track_id) AS track_count
FROM albums
JOIN tracks ON tracks.album_id = albums.album_id
GROUP BY albums.album_id, albums.title
''', label="Q6")
```

### Q7. How many listens does each track have?

```python

run_or_todo('''
SELECT tracks.title, COUNT(listens.listen_id) AS listen_count
FROM tracks
LEFT JOIN listens ON listens.track_id = tracks.track_id
GROUP BY tracks.track_id, tracks.title;
''', label="Q7")
```

### Q8. How many listens has each user made?

```python

run_or_todo('''
SELECT users.name, COUNT(listens.listen_id) AS listen_count
FROM users
LEFT JOIN listens ON listens.user_id = users.user_id
GROUP BY users.user_id, users.name;
''', label="Q8")
```

### Q9. Which tracks have been listened to more than 3 times?

```python

run_or_todo('''
SELECT tracks.title, COUNT(listens.listen_id) AS listen_count
FROM listens
Join tracks on listens.track_id = tracks.track_id
GROUP BY tracks.track_id, tracks.title
HAVING COUNT(listens.listen_id) > 3;
''', label="Q9")
```

### Q10. Which album has the most total listens?

```python

run_or_todo('''
SELECT albums.title, COUNT(listens.listen_id) AS listen_count
FROM listens
JOIN tracks ON listens.track_id = tracks.track_id
JOIN albums ON tracks.album_id = albums.album_id
GROUP BY albums.album_id, albums.title
ORDER BY listen_count DESC
LIMIT 1;
''', label="Q10")
```

### Q11. Which artist has the most total listens?

```python

run_or_todo('''
SELECT artists.name, COUNT(listens.listen_id) AS total_listens
FROM listens
JOIN tracks ON listens.track_id = tracks.track_id
JOIN albums ON tracks.album_id = albums.album_id
JOIN artists ON albums.artist_id = artists.artist_id
GROUP BY artists.artist_id, artists.name
ORDER BY total_listens DESC
LIMIT 1;
''', label="Q11")
```

### Q12. What is the average number of listens per user?

```python

run_or_todo('''
SELECT AVG(listen_count) AS avg_listens
FROM(
    SELECT users.user_id, COUNT(listens.listen_id) AS listen_count
    FROM users
    LEFT JOIN listens ON listens.user_id = users.user_id
    GROUP BY users.user_id
) AS Q12;
''', label="Q12")
```

### Q13. Which users have never listened to a track?

```python

run_or_todo('''
SELECT users.name
FROM users
LEFT JOIN listens ON listens.user_id = users.user_id
WHERE listens.user_id IS NULL;
''', label="Q13")
```

### Q14. Show the top 5 most-listened-to tracks.

```python

run_or_todo('''
SELECT tracks.title, COUNT(listens.listen_id) AS listen_count
FROM listens
JOIN tracks ON listens.track_id = tracks.track_id
GROUP BY tracks.track_id, tracks.title
ORDER BY listen_count DESC
LIMIT 5
''', label="Q14")
```
